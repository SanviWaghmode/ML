```python
def func(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)

def gradient_descent(starting_x, learning_rate, num_iterations):
    x = starting_x
    gd = [x]

    for i in range(num_iterations):
        grad = derivative(x)
        x = x - learning_rate * grad
        gd.append(x)

        print(f"Iteration {i+1}: x={x:.4f}, f(x)={func(x):.4f}")

    return x, gd

min_x, gd = gradient_descent(2, 0.1, 50)

print("Local minimum at x =", min_x)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
y_values  =[func(x) for x in gd]
sns.scatterplot(x = gd , y= y_values)
plt.title("Gradient Descent on (x+3)^2)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show();
```
